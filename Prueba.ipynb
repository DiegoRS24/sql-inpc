```python
import sqlite3, pandas as pd, re

conn = sqlite3.connect("data/inpc.db")

def ejecutar_archivo_sql(ruta):
    texto = open(ruta, encoding="utf-8").read()
    bloques = re.split(r"\n-- -{5,}\n", texto)
    for bloque in bloques:
        # Quitar comentarios de línea para que sqlite lo acepte
        limpio = "\n".join(l for l in bloque.splitlines() if not l.strip().startswith("--"))
        if not limpio.strip() or not ";" in limpio:
            continue
        # Mostrar el comentario (la pregunta) y el resultado
        comentarios = [l for l in bloque.splitlines() if l.strip().startswith("--")]
        print("\n".join(comentarios))
        try:
            display(pd.read_sql_query(limpio.split(";")[0] + ";", conn))
        except Exception as e:
            print("Error:", e)

ejecutar_archivo_sql("02_consultas_basicas.sql")
```
